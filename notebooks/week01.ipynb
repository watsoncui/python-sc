## 1.2 Read CSV & Basic Stats

```
df = pd.read_csv("../data/week01_sample.csv")
print(df)
mean_y = df["y"].mean()
print(f"Mean of y: {mean_y:.4f}")
```

## 1.3 Linear Fit & Plot

```
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

def linear(x, a, b):
    return a * x + b

popt, _ = curve_fit(linear, df["x"].values, df["y"].values)
print(f"Fit: y = {popt[0]:.4f}*x + {popt[1]:.4f}")

fig, ax = plt.subplots()
ax.scatter(df["x"], df["y"], label="data")
x_fit = np.linspace(0, 5, 100)
ax.plot(x_fit, linear(x_fit, *popt), "r-", label=f"fit: y={popt[0]:.2f}x+{popt[1]:.2f}")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_title("Week 01: CSV Data + Linear Fit")
ax.legend()
ax.grid(True, alpha=0.3)
plt.show()
```
